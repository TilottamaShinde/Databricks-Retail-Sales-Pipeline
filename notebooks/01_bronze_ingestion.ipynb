```python
catalog = "workspace"
schema = "default"
volume = "retail_sales_raw"

raw_path = f"/volumes/{catalog}/{schema}/{volume}"
print(raw_path)
```

```python
customers_df = (
    spark.read
    .format("csv")
    .option("header", "true")
    .option("inferSchema", "true")
    .load("/Volumes/workspace/default/retail_sales_raw/customers.csv")
)

display(customers_df)
```

```python
display(
    spark.sql("""
              LIST '/Volumes/workspace/default/retail_sales_raw'
              """)
)
```

```python
customers_df.printSchema()
```

```python
product_df = (
    spark.read
    .format("csv")
    .option("header","true")
    .option("inferSchema","true")
    .load("/Volumes/workspace/default/retail_sales_raw/products.csv")
)
display(product_df)
```

```python
product_df.printSchema()
```

```python
orders_df = (
    spark.read
    .format("csv")
    .option("header","true")
    .option("inferSchema","true")
    .load("/Volumes/workspace/default/retail_sales_raw/orders.csv")
)
display(orders_df)
```

```python
orders_df.printSchema()
```

Create a Bronze Schema

```python
spark.sql("CREATE SCHEMA IF NOT EXISTS workspace.bronze")
```

Save customers_df

```python
(
customers_df.write 
.format("delta")
.mode("overwrite") 
.saveAsTable("workspace.bronze.customers")
)
```

```python
(
    product_df.write
    .format("delta")
    .mode("overwrite")
    .saveAsTable("workspace.bronze.products")

)
```

```python
(
    orders_df.write
    .format("delta")
    .mode("overwrite")
    .saveAsTable("workspace.bronze.orders")
)
```
