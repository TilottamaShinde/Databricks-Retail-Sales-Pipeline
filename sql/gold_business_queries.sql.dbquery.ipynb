```sql
-- Monthly Revenue
SELECT * FROM
workspace.gold.monthly_revenue
ORDER BY year, month;


-- Product Revenue
SELECT * FROM 
workspace.gold.product_revenue
ORDER BY total_revenue DESC;

-- Revenue by Country
SELECT * FROM 
workspace.gold.country_revenue
ORDER BY total_revenue DESC;

-- Average Order Value
SELECT * FROM
workspace.gold.average_order_value;

-- Top Customers
SELECT * FROM
workspace.gold.top_customers
ORDER BY total_revenue DESC
LIMIT 10;
```
